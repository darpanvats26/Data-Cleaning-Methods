# **Data Cleaning**

```python
import pandas as pd
import numpy as np

# 1. LOAD DATASET

df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)

display(df.head())
```

```python
# Data Inspection
print("\n--- Dataset Information ---")
df.info()

print("\n--- Statistical Summary ---")
display(df.describe(include="all"))
```

## Checking Missing Values

```python
# Missing values in each column
print("Missing Values by Column:")
missing_values = df.isnull().sum()
display(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percentage.round(3)
})

missing_report
```

## Checking Duplicate Records

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Display duplicate records
display(df[df.duplicated(keep=False)])
```

```python
df = df.drop_duplicates()
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

## Checking Inconsistent Categorical Value

```python
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
#Standardize inconsistent values
text_columns = [
    "Employee_Name",
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

df["Department"] = df["Department"].str.title()
df["Gender"] = df["Gender"].str.title()
df["City"] = df["City"].str.title()
df["Work_Mode"] = df["Work_Mode"].str.title()
```

```python
print(df.dtypes)
```

```python
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)
print(df.dtypes)
```

## Handling Missing value in Numerical Column

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]
print(df[numeric_columns].isnull().sum())
```

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print(df[numeric_columns].isnull().sum())
```

## Handling Missing Categorical value

```python
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for column in categorical_columns:
    mode_value = df[column].mode()[0]
    df[column] = df[column].fillna(mode_value)

print(df[categorical_columns].isnull().sum())
```

```python
print("Missing Values After Cleaning")

display(df.isnull().sum())

print(
    "Total missing values:",
    df.isnull().sum().sum()
)
```

```python
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print("Final Dataset Information:")
df.info()

print("Final Dataset:")
display(df.head(10))
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
df["Joining_Date"] = df["Joining_Date"].dt.strftime("%Y-%m-%d")
```

```python
df.to_csv(
    "Cleaned_Company_Employee_Dataset_Day11.csv",
    index=False
)

print("Cleaned dataset exported successfully.")
```

## Final Comparision

```python
# Reload original dataset for comparison
original_df = pd.read_csv(
    "Day11_Messy_Company_Employee_Dataset.csv"
)

comparison = pd.DataFrame({
    "Measure": [
        "Number of Rows",
        "Number of Columns",
        "Total Missing Values",
        "Duplicate Rows"
    ],
    "Before Cleaning": [
        len(original_df),
        original_df.shape[1],
        original_df.isnull().sum().sum(),
        original_df.duplicated().sum()
    ],
    "After Cleaning": [
        len(df),
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

display(comparison)
```

### Summary

The company employee dataset was first loaded and inspected using Pandas to identify missing values, duplicate records, inconsistent entries, and incorrect data types. Text fields were cleaned by removing extra spaces and standardizing capitalization of categorical values such as department, gender, city, and work mode. Numeric columns were converted to appropriate numeric data types, while the joining date was converted to a proper date format. Missing numerical values were filled using median imputation, and missing categorical values were filled using the mode. Duplicate records were identified using duplicated() and removed using drop_duplicates(). Finally, the cleaned dataset was verified to ensure that there were no remaining missing values or duplicate records, and it was exported as a CSV file.
